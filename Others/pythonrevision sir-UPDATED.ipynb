## <span style = "color: red"> TUPLES </span>

A tuple is a collection similar to a Python list. The primary difference is that we cannot modify a tuple once it is created.

We create a tuple by placing items inside parentheses (). For example,

```python
hospital = ('37 military ', 'Korle bu', 'UGMC', 'Pantang psychiatric')

print(hospital)
```

## <span style =' color: blue'>Tuple Characteristics</span>
Tuples are:

1. Ordered - They maintain the order of elements.
2. Immutable - They cannot be changed after creation.
3. Allow duplicates - They can contain duplicate values.

## <span style =' color: blue'>Access Tuple Items</span>
Each item in a tuple is associated with a number, known as a index.

The index always starts from 0, meaning the first item of a tuple is at index 0, the second item is at index 1, and so on.

We use index numbers to access tuple items. For example,

```python
languages = ('Python', 'Swift', 'C++')

# access the first item
print(languages[0])   # Python

# access the third item
print(languages[2])   # C++

# access the last item
print(languages[-1])   # C++
```

## <span style = 'color:red'>Note: </span>Tuple Cannot be Modified
Python tuples are immutable (unchangeable). We cannot add, change, or delete items of a tuple.

If we try to modify a tuple, we will get an error. For example,

```python
biomed_level400 = ('Felix', 'Bethel', 'Vic', 'John', 'Camara', 'Dudzi')
print("Students in Biomed Level 400 are:", biomed_level400)


# try changing the second item to 'Melisa'
biomed_level400[1] = 'Melisa'

print('Updated List:', biomed_level400)

```

## <span style =' color: blue'>Python Tuple Length</span>

```python
medical_devices = ('dialysis', 'catheter', 'X-ray', 'MRI')
print('Total Items:', len(medical_devices )) 
       
```

## <span style = "color: red"> DICTIONARIES </span>
A Python dictionary is a collection of items, similar to lists and tuples. However, unlike lists and tuples, each item in a dictionary is a key-value pair (consisting of a key and a value).

We create a dictionary by placing key: value pairs inside curly brackets {}, separated by commas. For example,

```python
# creating a dictionary
patient_1 = {
    'last_name': 'Smith',
    'first_name': 'Bob',
    'date_of_birth': 'January 3, 1976',
    'number_of_children': 3
}



# printing the dictionary
print(patient_1)
```

### The patient_1 dictionary has four elements (key-value pairs), where 'last_name' is the key and 'Smith' is the value assigned to it and so on.

## <span style =' color: blue'>Valid and Invalid Dictionaries</span>

### Keys of a dictionary must be immutable
Immutable objects can't be changed once created. Some immutable objects in Python are integer, tuple and string.

```python
# valid dictionary
# integer as a key
my_dict1 = {1: "white blood cells", 2: "red blood cells", 3: "lymphocytes"}
print(my_dict1)

# valid dictionary
# tuple as a key
my_dict2 = {(1, 2): "jaundice", 3: "hepatitis C"}
print(my_dict2)

# invalid dictionary
# Error: using a list as a key is not allowed
my_dict3 = {1: "immune system", [1, 2]: "antigens"}
print(my_dict3)

# valid dictionary
# string as a key, list as a value
my_dict4 = {"Korle bu": ["Gynecology", "Urology", "Cardio-thoracic"]}
print(my_dict4)
```

### In this example, we have used integers, tuples, and strings as keys for the dictionaries.
### When we used a list as a key, an error message occurred due to the list's mutable nature.

## <span style = 'color:red'>Note: </span> Dictionary values can be of any data type, including mutable types like lists.

## <span style =' color: blue'>Accessing Dictionary Items</span>
We can access the value of a dictionary item by placing the key inside square brackets.

```python
patient_1 = {
    'last_name': 'Smith',
    'first_name': 'Bob',
    'date_of_birth': 'January 3, 1976',
    'number_of_children': 3
}

# access the value of keys
print(patient_1["last_name"])    
print(patient_1["first_name"])
```

## <span style =' color: blue'>Add Items to a Dictionary</span>
We can add an item to a dictionary by assigning a value to a new key. For example,

```python
patient_1 = {
    'last_name': 'Smith',
    'first_name': 'Bob',
    'date_of_birth': 'January 3, 1976',
    'number_of_children': 3
}



# add an item with "weight" as key and "120 kg" as its value
patient_1["weight"] = "120 kg"

print(patient_1)
```

## <span style =' color: blue'>Change Dictionary Items</span>
Python dictionaries are mutable (changeable). We can change the value of a dictionary element by referring to its key. For example,

```python
patient_1 = {
    'last_name': 'Smith',
    'first_name': 'Bob',
    'date_of_birth': 'January 3, 1976',
    'number_of_children': 3
}



# replace Smith with "Morgan" as value for the key last_name 
patient_1["last_name"] = "Morgan"

print(patient_1)
```

## <span style =' color: blue'>Remove Dictionary Items</span>
We can use the del statement to remove an element from a dictionary. For example,

```python
patient_1 = {
    'last_name': 'Smith',
    'first_name': 'Bob',
    'date_of_birth': 'January 3, 1976',
    'number_of_children': 3
}

# delete item having "date_of_birth" key
del patient_1['date_of_birth']

print(patient_1)
```

### If we need to remove all items from a dictionary at once, we can use the clear() method.

```python
patient_1 = {
    'last_name': 'Smith',
    'first_name': 'Bob',
    'date_of_birth': 'January 3, 1976',
    'number_of_children': 3
}


# clear the dictionary
patient_1.clear()

print(patient_1)  
```

## <span style =' color: blue'>The return Statement</span>

We return a value from the function using the return statement.

```python
# function definition
def find_square(num):
    result = num * num
    return result

# function call

square = find_square(3)

print('Square:', square)
```
